```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf   ## Imports for them

# Example: Create a simple time series (or replace with your own data)
np.random.seed(0)
data = np.cumsum(np.random.randn(100))  # Random walk

# Convert to pandas Series
ts = pd.Series(data)

# Plot ACF and PACF
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plot_acf(ts, lags=20, ax=plt.gca(), title='Autocorrelation (ACF)')       ## ACF

plt.subplot(1, 2, 2)
plot_pacf(ts, lags=20, ax=plt.gca(), title='Partial Autocorrelation (PACF)', method='ywm')   ## PACF

plt.tight_layout()
plt.show()
```
