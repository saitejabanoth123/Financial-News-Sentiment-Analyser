```python
## Selection Sorting

def select_sort(arr):
    n=len(arr)
    for i in range(n):
        min_idx=i
        for j in range(i+1,n):
            if arr[j]<arr[min_idx]:
                min_idx=j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
```

```python
select_sort([13,46,24,52,20,9])   ## Time Complexity = O(N**2) Best, Worst and Average all 
```

```python
## Bubble Sorting

def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Last i elements are already in place
        for j in range(0, n - i - 1):
            # Swap if the element found is greater than the next
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
    
```

```python
bubble_sort([13,46,24,52,20,9])    ## Time Complexity: Best----> O(N)   Average and Worst------> O(N**2)
```

```python
## Insertion Sorting

def insertion_sort(arr):
    # Traverse from 1 to len(arr)
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key    ## to not leave blank space which is filled with repeated numbers then 
    
    return arr
```

```python
insertion_sort([13,46,24,52,20,9])    ## Time Complexity: Best----> O(N)   Average and Worst------> O(N**2)
```
