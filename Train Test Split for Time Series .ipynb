```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ts=pd.read_csv('AirPassengers.csv', parse_dates=['Month'],index_col='Month')
```

```python
ts.shape
```

```python
ts.head(144)
```

```python
## First Method
```

```python
## 80% Train 20% Split

train_size=int(len(ts)*0.8)     # train → past data to fit the model 
                                # test → future data to evaluate predictions
```

```python
train=ts[:train_size]
test=ts[train_size:]
```

```python
## Second Method
```

```python
split_data='1960-01-01'
```

```python
train1=ts[ts.index<split_data]
test1=ts[ts.index>split_data]
```

```python
plt.plot(train, label='Train')
plt.plot(test, label='Test')

plt.xlabel('Year')
plt.ylabel('No. of Passengers')

plt.title('1st Method')

plt.legend()
plt.show()
```

```python
plt.plot(train1, label='Train')
plt.plot(test1, label='Test')

plt.xlabel('Year')
plt.ylabel('No. of Passengers')

plt.title('2nd Method')

plt.legend()
plt.show()
```
